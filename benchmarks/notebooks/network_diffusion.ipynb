# Virus on a network

The bundled virus-on-network model on Erdos-Renyi graphs of growing size. Agents are fixed to nodes; the work per step is the neighbourhood walk of every infected node.

```python
import json
import time

import pandas as pd


def bench(label, make_model, steps, agents_of=lambda m: len(m.agents)):
    """Build the model, time ``steps`` steps, print a RESULT line, return a row."""
    model = make_model()
    n = agents_of(model)
    t0 = time.perf_counter()
    model.run_for(steps)
    per_step = (time.perf_counter() - t0) / steps
    row = {
        "label": label,
        "agents": n,
        "ms_per_step": round(per_step * 1e3, 2),
        "us_per_agent_step": round(per_step / max(n, 1) * 1e6, 2),
    }
    print("RESULT " + json.dumps(row))
    return row
```

```python
from modelly.examples import VirusOnNetwork


def virus(nodes):
    return lambda: VirusOnNetwork(num_nodes=nodes, avg_node_degree=6,
                                  initial_outbreak_size=max(1, nodes // 100), rng=1)
```

```python
rows = []
rows.append(bench("2k nodes", virus(2000), steps=10))
rows.append(bench("8k nodes", virus(8000), steps=10))
rows.append(bench("20k nodes", virus(20000), steps=5))
```

```python
pd.DataFrame(rows).set_index('label')
```
